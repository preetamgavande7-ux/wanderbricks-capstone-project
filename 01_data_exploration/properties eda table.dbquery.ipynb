```sql
use databricks_wanderbricks_dataset_dais_2025.wanderbricks;

select * from properties limit 10;


show tables;

describe properties;

-- total count of property record
select count(*) from properties;

-- how many unique host are there
select count(distinct host_id) as unique_hosts from properties;

-- are there missing values
select
count(*) as total_rows,
sum(case when property_id is null then 1 else 0 end) as missing_property_id,
sum(case when host_id is null then 1 else 0 end) as missing_host_id,
sum(case when city_id is null then 1 else 0 end) as missing_city_id,
sum(case when base_price is null then 1 else 0 end) as missing_price,
sum(case when property_type is null then 1 else 0 end) as missing_property_type, 
sum(case when max_guests is null then 1 else 0 end) as missing_max_guests,
sum(case when bedrooms is null then 1 else 0 end) as missing_bedrooms,
sum(case when bathrooms is null then 1 else 0 end) as missing_bathrooms,
sum(case when created_at is null then 1 else 0 end) as missing_created_at
from properties;


-- total city count
select count(distinct(city_id)) from properties;

-- what type of total property type exist
SELECT DISTINCT property_type
FROM properties
ORDER BY property_type;

-- minimum base price of properties by property_type
SELECT property_type, MIN(base_price) AS min_base_price
FROM properties
GROUP BY property_type;

SELECT city_id, MIN(base_price)AS min_base_price
FROM properties
GROUP BY city_id;

-- maximum base price of properties by property_type
SELECT property_type, city_id, max(base_price) AS max_base_price
FROM properties
GROUP BY property_type, city_id;

-- which cities have most properties
select city_id, count(*) as property_count from properties group by city_id order by property_count desc;

-- min,max,avg
select min(base_price) as min_price,
max(base_price) as max_price,
avg(base_price) as avg_price from properties;


select property_type, avg(base_price) as avg_price
from properties
group by property_type
order by avg_price desc;

SELECT property_type, count(*) AS property_count
FROM properties
GROUP BY property_type
order by property_count desc;

-- creating view using city and property table
create or replace temp view properties_city_view as 
select p.*, c.city, c.country from properties p
join cities c
on p.city_id = c.city_id;

select * from properties_city_view limit 10;

-- property count by city 
select city, count(*) as property_count
from properties_city_view
group by city
order by property_count desc;


/*
=== Properties EDA Insights ===
1. The properties table contains 50000 property record.
2. These properties are managed by 9939 unique hosts.
3. Properties are available across 39 cities.
4. The platform contains multiple properties type such as histrocial places, Ski Resort, Summer, Gateways, Urban Year Round.
5. Avg base price are very similar across all property type.
6. Property type does not show a major difference in listing price.
7. Property listings are also evenly distributed across property type. almost 12k (25%) for each property type.
8. Singapore as the highest listed properties and there is a major gap between the first and second city.
*/
```
