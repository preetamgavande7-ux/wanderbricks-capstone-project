```sql
use databricks_wanderbricks_dataset_dais_2025.wanderbricks;

show tables;

-- Exploring Users Tables

-- 1. what does the table contain? 

-- 2. what does one row represent?
-- 3. what columns are available ?
-- 4. how many rows are there?


select * from users limit 10;
describe users;
-- count of users
select count(*) as total_rows, count(distinct user_id) as unique_users from users;


-- describe
DESCRIBE amenities;
DESCRIBE bookings;
DESCRIBE cities;
DESCRIBE employees;
DESCRIBE hosts;
DESCRIBE properties;
DESCRIBE property_amenities;
DESCRIBE property_images;
DESCRIBE users;


-- users table

-- 1. contains info about wanderbricks users.
-- 2. one row repesents one user.
--3. columns tells us what info we know about each user
--4. there are 100000 rows
-- 5. null values in email and company_name column 
-- 6. users can be either indiviual or business. 
-- 7. we can study user data by country, account_type, and signup data

-- what is the distribtuion of users by country
select country, count(*) as count
 from users group by country
  order by count desc;


-- what is the distribution of users by user type?
select user_type, count(*) as ucount
from users 
group by user_type
order by ucount desc;

-- any missing values
select * from users 
where user_id is null;

SELECT
    SUM(CASE WHEN user_id IS NULL THEN 1 ELSE 0 END) AS missing_user_id,
    SUM(CASE WHEN email IS NULL THEN 1 ELSE 0 END) AS missing_email,
    SUM(CASE WHEN name IS NULL THEN 1 ELSE 0 END) AS missing_name,
    SUM(CASE WHEN country IS NULL THEN 1 ELSE 0 END) AS missing_country,
    SUM(CASE WHEN user_type IS NULL THEN 1 ELSE 0 END) AS missing_user_type,
    SUM(CASE WHEN created_at IS NULL THEN 1 ELSE 0 END) AS missing_created_at,
    SUM(CASE WHEN is_business IS NULL THEN 1 ELSE 0 END) AS missing_is_business,
    SUM(CASE WHEN company_name IS NULL THEN 1 ELSE 0 END) AS missing_company_name
FROM users;
-- what does our user base looks like ?
select country, user_type,  count(*) as ucount
from users
group by country, user_type
order by country;
-- when did users join platform (earliest sign up, latest signup, user joining by month/year)
select min(created_at) as earliestsignup, max(created_at) as latestsignup from users;

select month(created_at) as signup_month, count(*) as tusers
from users
group by month(created_at)
order by signup_month;
```
